```python
# =======================================
# 1.- ENTORNO BÁSICO DE ANÁLISIS EN COLAB
# =======================================

import numpy as np         # Librería para computación numérica
import pandas as pd        # Librería para manipulación de datos tabulares
import matplotlib.pyplot as plt  # Para visualización de gráficos básicos
import seaborn as sns      # Biblioteca complementaria para visualización

# Estilo visual predeterminado
sns.set(style="whitegrid")
```

```python
# ==========================================
# 2.- CRECIMIENTO POBLACIONAL SIMULADO
# ==========================================

# Simulamos crecimiento con una función exponencial
anios = np.arange(2000, 2025)  # Años desde 2000 hasta 2024
poblacion = 50000 * np.exp(0.03 * (anios - 2000))  # Crecimiento anual del 3%

# Construcción del DataFrame
df_poblacion = pd.DataFrame({
    'Año': anios,
    'Población': poblacion.astype(int)
})

df_poblacion.head()
```

```python
# 3.- VISUALIZACIÓN DEL CRECIMIENTO
plt.figure(figsize=(10, 5))
sns.lineplot(data=df_poblacion, x='Año', y='Población', marker='o', color="purple")
plt.title("Crecimiento Poblacional Simulado (2000–2024)")
plt.xlabel("Año")
plt.ylabel("Número de Habitantes")
plt.grid(True)
plt.show()
```

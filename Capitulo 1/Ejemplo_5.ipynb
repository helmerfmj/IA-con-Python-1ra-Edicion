```python
# ================================================
# 1.- DEFINICIÓN DE UNA FUNCIÓN BÁSICA EN GOOGLE COLAB
# ================================================

def saludar(nombre):
    """
    Esta función imprime un saludo personalizado.
    Parámetros:
    nombre (str): nombre de la persona.
    """
    print(f"Hola, {nombre}. Bienvenido al mundo de la IA.")

# Llamada a la función
saludar("Helmer")
```

```python
# ===========================================
# 2.- FUNCIÓN QUE RETORNA EL DOBLE DE UN NÚMERO
# ===========================================

def doble(x):
    """
    Devuelve el doble del valor ingresado.
    """
    return 2 * x

# Uso de la función
resultado = doble(5)
print(f"El resultado es: {resultado}")
```

```python
# ======================================
# 3.- FUNCIÓN QUE APLICA OTRA FUNCIÓN
# ======================================

def aplicar_operacion(funcion, valor):
    """
    Aplica una operación matemática a un valor dado.
    """
    return funcion(valor)

# Uso con funciones lambda
resultado = aplicar_operacion(lambda x: x**2 + 3, 4)
print(f"Resultado: {resultado}")
```

```python
# ==================================================
# 4.- DECORADOR BÁSICO PARA ANUNCIAR LA EJECUCIÓN
# ==================================================

def anunciar(func):
    """
    Decorador que imprime un mensaje antes y después
    de ejecutar la función original.
    """
    def envoltura(*args, **kwargs):
        print(">>> Iniciando ejecución...")
        resultado = func(*args, **kwargs)
        print(">>> Ejecución finalizada.")
        return resultado
    return envoltura

# Aplicación del decorador con @
@anunciar
def entrenar_modelo():
    print("Entrenando modelo de IA...")

# Llamada a la función decorada
entrenar_modelo()
```

```python
# ============================================
# 5.- GENERADOR DE LA SECUENCIA DE FIBONACCI
# ============================================

def fibonacci(n):
    """
    Generador de los primeros n números de Fibonacci.
    """
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b

# Iteración sobre el generador
for numero in fibonacci(8):
    print(numero, end=" ")
```

```python
# =======================================
# ENTORNO BÁSICO DE ANÁLISIS EN COLAB
# =======================================

import numpy as np         # Librería para computación numérica
import pandas as pd        # Librería para manipulación de datos tabulares
import matplotlib.pyplot as plt  # Para visualización de gráficos básicos
import seaborn as sns      # Biblioteca complementaria para visualización

# Estilo visual predeterminado
sns.set(style="whitegrid")
```

```python
# ==========================================
# SIMULACIÓN DE DATOS DE VENTAS DIARIAS
# ==========================================

np.random.seed(42)  # Fijamos semilla para reproducibilidad

dias = pd.date_range(start='2024-01-01', periods=30)
ventas = np.random.normal(loc=250.0, scale=50.0, size=30).round(2)
clientes = np.random.poisson(lam=30, size=30)

# Construcción del DataFrame
df = pd.DataFrame({
    'Fecha': dias,
    'Ventas ($)': ventas,
    'Clientes': clientes
})

df.head()
```

```python
# ==========================================
# ANÁLISIS ESTADÍSTICO BÁSICO
# ==========================================

# Con pandas
df.describe()
```

```python
# Con NumPy (acceso individual)
media_ventas = np.mean(df['Ventas ($)'])
mediana_ventas = np.median(df['Ventas ($)'])
desviacion = np.std(df['Ventas ($)'])

print(f"Media: {media_ventas:.2f}")
print(f"Mediana: {mediana_ventas:.2f}")
print(f"Desviación estándar: {desviacion:.2f}")
```

```python
# ==========================================
# VISUALIZACIÓN DE LA DISTRIBUCIÓN
# ==========================================

plt.figure(figsize=(10, 5))
sns.histplot(df['Ventas ($)'], kde=True, bins=10, color="skyblue")
plt.title("Distribución de Ventas Diarias")
plt.xlabel("Monto en dólares")
plt.ylabel("Frecuencia")
plt.show()
```

```python
# RELACIÓN ENTRE CLIENTES Y VENTAS
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Clientes', y='Ventas ($)', color="darkgreen")
plt.title("Relación entre Clientes y Ventas")
plt.xlabel("Número de clientes")
plt.ylabel("Monto vendido ($)")
plt.show()
```

```python
# CORRELACIÓN DE PEARSON
correlacion = df['Clientes'].corr(df['Ventas ($)'])
print(f"Correlación Pearson: {correlacion:.3f}")
```

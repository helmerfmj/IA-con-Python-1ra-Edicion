```python
# =======================================
# 1.- ENTORNO BÁSICO DE ANÁLISIS EN COLAB
# =======================================

import numpy as np         # Librería para computación numérica
import pandas as pd        # Librería para manipulación de datos tabulares
import matplotlib.pyplot as plt  # Para visualización de gráficos básicos
import seaborn as sns      # Biblioteca complementaria para visualización

# Estilo visual predeterminado
sns.set(style="whitegrid")
```

```python
# ==========================================
# 2.- INGRESOS Y GASTOS EN HOGARES
# ==========================================

# Simulación de datos
np.random.seed(10)
n = 100
ingresos = np.random.normal(loc=1500, scale=300, size=n)
gastos = np.random.normal(loc=1350, scale=350, size=n)

df_hogares = pd.DataFrame({
    'Ingresos': ingresos.round(2),
    'Gastos': gastos.round(2)
})
df_hogares['Ahorro'] = df_hogares['Ingresos'] - df_hogares['Gastos']

df_hogares.head()
```

```python
# 3.- ESTADÍSTICAS DE AHORRO
print("Estadísticas de ahorro mensual:")
print(df_hogares['Ahorro'].describe())

# Hogares con déficit (ahorro negativo)
deficit = df_hogares[df_hogares['Ahorro'] < 0]
print(f"\nNúmero de hogares con déficit: {len(deficit)}")
```

```python
# 4.- VISUALIZACIÓN DE LA DISTRIBUCIÓN DEL AHORRO
plt.figure(figsize=(8, 5))
sns.histplot(df_hogares['Ahorro'], kde=True, color="salmon", bins=15)
plt.axvline(0, color='black', linestyle='--')
plt.title("Distribución del Ahorro Mensual en Hogares")
plt.xlabel("Ahorro en dólares")
plt.ylabel("Frecuencia")
plt.show()
```

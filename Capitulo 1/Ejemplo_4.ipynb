```python
# =======================================
# 1.- ENTORNO BÁSICO DE ANÁLISIS EN COLAB
# =======================================

import numpy as np         # Librería para computación numérica
import pandas as pd        # Librería para manipulación de datos tabulares
import matplotlib.pyplot as plt  # Para visualización de gráficos básicos
import seaborn as sns      # Biblioteca complementaria para visualización

# Estilo visual predeterminado
sns.set(style="whitegrid")
```

```python
# ==========================================
# 2.- ESTUDIO SOCIOECONÓMICO SIMULADO
# ==========================================

np.random.seed(20)
personas = 200

edad = np.random.randint(18, 65, size=personas)
educacion = np.random.randint(8, 20, size=personas)  # Años de estudio
ingresos = educacion * 120 + np.random.normal(0, 500, size=personas)  # Relación directa con ruido

df_socio = pd.DataFrame({
    'Edad': edad,
    'Años de Educación': educacion,
    'Ingresos Mensuales': ingresos.round(2)
})
df_socio.head()
```

```python
# 3.- MATRIZ DE CORRELACIÓN
cor_matrix = df_socio.corr()
print("Matriz de correlación:")
print(cor_matrix)
```

```python
# VISUALIZACIÓN DE LA MATRIZ DE CORRELACIÓN
plt.figure(figsize=(8, 6))
sns.heatmap(cor_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Matriz de Correlación entre Variables Socioeconómicas")
plt.show()
```
